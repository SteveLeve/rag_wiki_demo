# Advanced Techniques

Six ways to make retrieval better than plain cosine similarity over fixed-size chunks, plus two modern additions. Each notebook isolates one technique, measures it against the foundation baseline, and records the run in the `experiments` table so results are comparable across notebooks.

## Notebooks

| Notebook | What it covers |
|---|---|
| [`05-reranking.ipynb`](./05-reranking.ipynb) | Cross-encoder second pass over the top-N vector hits. |
| [`06-query-expansion.ipynb`](./06-query-expansion.ipynb) | Generate query variants with the local LLM, retrieve on each, merge. |
| [`07-hybrid-search.ipynb`](./07-hybrid-search.ipynb) | Dense vectors + BM25 keyword search, fused with reciprocal rank fusion. |
| [`08-semantic-chunking-and-metadata.ipynb`](./08-semantic-chunking-and-metadata.ipynb) | Split on meaning rather than character count; filter on metadata. |
| [`09-citation-tracking.ipynb`](./09-citation-tracking.ipynb) | Make the generated answer point back at the chunks that produced it. |
| [`10-combined-advanced-rag.ipynb`](./10-combined-advanced-rag.ipynb) | Capstone: all of the above in one configurable pipeline. |
| [`11-contextual-retrieval.ipynb`](./11-contextual-retrieval.ipynb) | Prepend LLM-written situating context to each chunk before embedding. |
| [`12-modern-reranking.ipynb`](./12-modern-reranking.ipynb) | Cross-encoder vs. LLM-as-reranker vs. none, with latency measured. |

**Prerequisite:** run `foundation/02-rag-postgresql-persistent.ipynb` first so an embedding model is registered. Any notebook here can then be run on its own — `ragkit.registry.ensure_embeddings()` generates what is missing.
