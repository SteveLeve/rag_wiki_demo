# Evaluation Lab

Techniques are worthless without measurement. This tier builds the ground truth, defines the metrics, establishes a baseline, and visualizes how variants compare — so every claim made in `advanced-techniques/` can be checked.

## Notebooks

| Notebook | What it covers |
|---|---|
| [`01-create-ground-truth-human-in-loop.ipynb`](./01-create-ground-truth-human-in-loop.ipynb) | Generate candidate questions, then curate them by hand. |
| [`02-evaluation-metrics-framework.ipynb`](./02-evaluation-metrics-framework.ipynb) | Precision@K, Recall@K, MRR, NDCG, and generation-quality metrics. |
| [`03-baseline-and-comparison.ipynb`](./03-baseline-and-comparison.ipynb) | Baseline vs. variants, with significance testing. |
| [`04-experiment-dashboard.ipynb`](./04-experiment-dashboard.ipynb) | Timeline and comparison views over the experiments table. |
| [`05-supplemental-embedding-analysis.ipynb`](./05-supplemental-embedding-analysis.ipynb) | Retrieval quality by query type; debugging poor retrievals. |

**Prerequisite:** run `foundation/02-rag-postgresql-persistent.ipynb` first. Notebook 01 populates `evaluation_groundtruth`, which 02-05 read.
