## 05 — Matryoshka Dimensions

`nomic-embed-text` is trained so that the first N components of its vector are
themselves a usable embedding. `ollama.embed(..., dimensions=256)` truncates and
re-normalizes, letting you trade index size against quality with one parameter
rather than a second model.

Notebook 05 measures that trade instead of assuming it: similarity margins at
128 / 256 / 512 / 768, storage per width, and NDCG against a labelled set.
